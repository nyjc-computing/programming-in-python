# Lesson 13: Abstraction with Modules

In this lesson, we will explore how Python modules help us achieve abstraction by organizing and reusing code. Modules allow us to group related functions, classes, and variables into separate files, making our programs more modular and maintainable.

Before diving into modules, let's revisit some examples of abstraction we've used so far: loops and functions.

## Examples of Abstraction

### Loops
Loops allow us to repeat a block of code multiple times without writing it repeatedly. For example:

```python
for i in range(5):
    print(f"This is iteration {i}")
```

### Functions
Functions encapsulate chunks of code into reusable blocks. For example:

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
print(greet("Bob"))
```

Both loops and functions help us reduce repetition and make our code more readable and maintainable.

## Introduction to Modules

Modules take abstraction a step further by allowing us to organize related code into separate files. This makes it easier to reuse code across multiple programs and keeps our projects organized.

### Importing Modules
To use a module, we import it into our program using the `import` statement. For example:
```python
import math

print(math.sqrt(16))  # Outputs: 4.0
```

### The `__name__` Variable
Every Python module has a special variable called `__name__`. When a module is run directly, `__name__` is set to `'__main__'`. When it is imported, `__name__` is set to the module's name. This allows us to write code that only runs when the module is executed directly.

### Aliasing
We can use aliases to give modules or functions shorter names for convenience. For example:
```python
import numpy as np

array = np.array([1, 2, 3])
print(array)
```

Even though the above functions are not exhaustive (by themselves, they will not be enough to create a game), you can already see it is going to be quite disorganized as we continue to add more functions.

We can bring clarity to code organization by reorganizing it as follows:

- `text.py`: contains in-game text
- `interaction.py`: contains functions for interacting with users and validating choices

With this reorganization, `main.py` might look like this:

### `__name__ == "__main__"`

While developing the game and testing it, we won't always be calling `run()` from `main.py`, because who wants to go through the intro text and enter player name *fifty times* right?

Often we might run shorter code within `game.py` to test portions of game functionality, e.g.

````python
# game.py

# Function definitions go here
# ...

# Testing the game logic
load_game()
game.execute_move("pickup")
# Check that the item picked up appears in inventory
# ...
````

Unfortunately, when we run `import game` in `main.py`, all this testing code is also executed, which is not what we intended. We would have to remove this testing code to run the game, then put it back when we want to do testing ... this much manual editing is not ideal.

#### Checking if the module is the entry point

When we run a program, we run it from a file, which can import code from other files. The file from which the program is run is called the **entry point** (usually `main.py`, but e.g. in the case of testing might be `game.py`).

In the entry point, the special `__name__` variable has the string value `"__main__"`, while other modules have their module name instead.

This is why you often see the following code:

```python
if __name__ == "__main__":
    # ...
```

This means that the code is meant to be executed only if run as the entrypoint. So we can rewrite `game.py` as follows:

```python
# game.py

# Function definitions go here
# ...

if __name__ == "__main__":
    # Testing the game logic
    load_game()
    game.execute_move("pickup")
    # Check that the item picked up appears in inventory
    # ...
```

This allows `game.py` to be imported by `main.py` without running the test code, while also allowing `game.py` to be run for testing.
