# Lesson 11: Handling CSV files with `csv`

## Introduction to CSV format

CSV (Comma-Separated Values) is a simple file format used to store tabular data, such as a spreadsheet or database. Each line in a CSV file represents a row, and each value is separated by a comma.

### CSV Format Rules
- Each row is a new line in the file.
- Values in a row are separated by commas. There should be no spaces between the values and comma.
- Strings containing commas or special characters must be enclosed in double quotes (`"`).
- Double quotes within a quoted string are escaped by doubling them (`""`).

For example, `addressbook.csv` might contain the following data:

    John Doe,30,"123, Elm Street"
    Jane Smith,25,456 Oak Avenue

### Headers
A CSV file may include a header row as the first line, which contains the names of the columns. This makes it easier to interpret the data.

For example, `addressbook.csv` with a header would look like this:

    Name,Age,Address
    John Doe,30,"123, Elm Street"
    Jane Smith,25,456 Oak Avenue

## `reader` and `writer` for CSV without header

The `csv.reader` and `csv.writer` classes are used to read and write CSV files without headers. Below is an example of how to use them.

### Reading a CSV file without headers
```python
import csv

# Reading a CSV file
with open('data.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)  # Each row is a list of strings
```

### Writing to a CSV file without headers
```python
import csv

# Writing to a CSV file using writerow()
with open('output.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["John", 30, "123 Elm Street"])  # Write a single row
    writer.writerow(["Jane", 25, "456 Oak Avenue"])

# Writing to a CSV file using writerows()
data = [["Alice", 28, "789 Pine Road"], ["Bob", 35, "101 Maple Lane"]]
with open('output_bulk.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)  # Write multiple rows
```

In the above examples:
- `writerow()` writes a single row to the file.
- `writerows()` writes multiple rows from a list of lists.

**Note:** Ensure the file is opened in `newline=''` mode to avoid extra blank lines in the output.

## `DictReader` and `DictWriter` for CSV with header

The `csv.DictReader` and `csv.DictWriter` classes are used to read and write CSV files with headers. Below is an example of how to use them.

### Reading a CSV file with headers
```python
import csv

# Reading a CSV file with headers
with open('data_with_headers.csv', mode='r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)  # Each row is a dictionary, with keys as column headers, and string values
```

### Writing to a CSV file with headers
```python
import csv

# Writing to a CSV file with headers using writerow()
with open('output_with_headers.csv', mode='w', newline='') as file:
    fieldnames = ["Name", "Age", "Address"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()  # Write the header row
    writer.writerow({"Name": "John", "Age": 30, "Address": "123 Elm Street"})  # Write a single row
    writer.writerow({"Name": "Jane", "Age": 25, "Address": "456 Oak Avenue"})

# Writing to a CSV file with headers using writerows()
data = [
    {"Name": "Alice", "Age": 28, "Address": "789 Pine Road"},
    {"Name": "Bob", "Age": 35, "Address": "101 Maple Lane"}
]
with open('output_bulk_with_headers.csv', mode='w', newline='') as file:
    fieldnames = ["Name", "Age", "Address"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()  # Write the header row
    writer.writerows(data)  # Write multiple rows
```

In the above examples:
- `writerow()` writes a single row as a dictionary.
- `writerows()` writes multiple rows from a list of dictionaries.

**Note:** The `fieldnames` parameter must match the keys in the dictionaries being written.
