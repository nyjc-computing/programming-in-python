# Lesson 11: Handling CSV files with `csv`

In lesson 9a, you learned how to read and write text files. But what if your data is structured in a table format, like a spreadsheet or a database?

This is where **CSV** files come in. CSV (Comma-Separated Values) is a simple file format used to store tabular data. Each line represents a row, and values are separated by commas.

### Why do we need the `csv` module?

You might wonder: can't I just use `split(',')` to parse CSV files? Let's see why that's a bad idea:

````python
# The problem: manual CSV parsing is error-prone
line = '"Smith, John",30,"123, Elm Street","New York, NY"'
parts = line.split(',')
print("Split result:", parts)
print("Problem: Can't tell which commas are separators!")
````

The `split(',')` approach breaks when data contains commas within quoted strings. The `csv` module handles all these edge cases correctly.

### Crash course: CSV format rules

Before we dive in, let's understand the CSV format:

- **Each row** is a new line in the file
- **Values** are separated by commas (no spaces after commas)
- **Headers**: The first row may contain column names
- **Quoting**: Strings containing commas, quotes, or newlines must be enclosed in double quotes (`"`)
- **Escaping quotes**: Double quotes within a quoted string are escaped by doubling them (`""`)

### Examples:

**Without headers:**
```
Apple,1.50,Fruit
Banana,0.80,Fruit
```

**With headers:**
```
Name,Age,City
Alice,25,New York
Bob,30,London
```

**With quoted fields (containing commas):**
```
Name,Address
"Smith, John","123 Main St, Apt 4B"
```

## Reading CSV files with `csv.reader`

Let's start with the simplest case: reading CSV files **without headers** using `csv.reader`.

### Basic `csv.reader` usage

Run the code below to see how `csv.reader` works:

````python
import csv

# Read products_no_header.csv
with open('data_files/products_no_header.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
        print('Row type:', type(row))
        print('---')
````

### Investigation: What type is each `row`?

Notice that:
1. Each `row` is a **list** (not a tuple or dict)
2. All values in the row are **strings** (even the price `1.50`)
3. You access values by **index** (e.g., `row[0]`, `row[1]`)

This means you need to **convert data types** yourself:

````python
# Notice that all values are strings! We need to convert types.
with open('data_files/products_no_header.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        name = row[0]
        price = float(row[1])  # Convert to float
        category = row[2]
        print(name + ': $' + str(price) + ' (' + category + ')')
````

## Writing CSV files with `csv.writer`

Now let's learn how to **write** CSV files using `csv.writer`.

### Writing simple data

Run the code below to create a new CSV file:

````python
import csv

data = [
    ["Apple", 1.50, "Fruit"],
    ["Banana", 0.80, "Fruit"],
    ["Carrot", 1.20, "Vegetable"]
]

with open('output_simple.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    for row in data:
        writer.writerow(row)

print("File created! Check the directory for output_simple.csv")
````

On Windows, without `newline=''`, you get double-spaced lines! This is because Python adds `\r\n` and the CSV module adds another `\n`. The `newline=''` parameter tells Python to let the CSV module handle newlines.

**Always use `newline=''` when working with CSV files!**
</details>

### Writing multiple rows at once

For efficiency, you can write multiple rows using `writerows()`:

````python
import csv

data = [
    ["Apple", 1.50, "Fruit"],
    ["Banana", 0.80, "Fruit"],
    ["Carrot", 1.20, "Vegetable"]
]

with open('output_bulk.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(data)  # Write all rows at once

print("Bulk write complete!")
````

## Reading CSV files with headers - `csv.DictReader`

When CSV files have headers, using `csv.DictReader` is often more convenient than `csv.reader`.

### Basic `csv.DictReader` usage

Notice the difference from `csv.reader`:

````python
import csv

# Now let's work with files that have headers
with open('data_files/simple_data.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
        print('Row type:', type(row))
        print('---')
````

### Investigation: List vs Dictionary

Notice that:
1. Each `row` is now a **dictionary** (not a list)
2. You access values by **column name** (e.g., `row['Name']`, `row['Age']`)
3. This is more **readable** and less error-prone than using indexes

Try accessing a value by column name:

````python
with open('data_files/simple_data.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        name = row['Name']
        age = int(row['Age'])  # Remember to convert types!
        city = row['City']
        print(name, 'is', str(age), 'years old and lives in', city)
````

## Writing CSV files with headers - `csv.DictWriter`

To write CSV files with headers, use `csv.DictWriter`.

### Writing with headers

Notice that you must specify the `fieldnames` parameter:

````python
import csv

data = [
    {"Name": "Alice", "Age": 25, "City": "New York"},
    {"Name": "Bob", "Age": 30, "City": "London"}
]

with open('output_with_headers.csv', 'w', newline='') as f:
    fieldnames = ["Name", "Age", "City"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    
    writer.writeheader()  # Don't forget this!
    writer.writerows(data)

print("File created with headers!")
````

## Error Handling and Troubleshooting

When working with files, things can go wrong. Let's learn how to handle errors gracefully (reference Lesson 10).

### File Not Found Error

What happens if you try to open a file that doesn't exist?

````python
import csv

try:
    with open('nonexistent.csv', 'r') as f:
        reader = csv.reader(f)
        for row in reader:
            print(row)
except FileNotFoundError as e:
    print('Error: File not found -', e)
````

### Handling Malformed CSV Data

What if CSV data has missing fields or unexpected values?

````python
# students_with_errors.csv has some rows with missing values
with open('data_files/students_with_errors.csv', 'r') as f:
    reader = csv.DictReader(f)
    rows = list(reader)
    for i in range(len(rows)):
        row = rows[i]
        row_num = i + 1
        print('Row', str(row_num) + ':', row)
        
        # Check for missing values
        if 'Age' not in row or not row['Age']:
            print('  Warning: Missing Age in row', row_num)
        if 'Grade' not in row or not row['Grade']:
            print('  Warning: Missing Grade in row', row_num)
````

### Handling Quoted Fields with Commas

The `csv` module correctly handles quoted fields containing commas:

````python
# books_malformed.csv has quoted fields with commas
with open('data_files/books_malformed.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print('Title:', row[0])
        print('Author:', row[1])  # Notice the comma in author name!
        print('Year:', row[2])
        print('---')
````

### Troubleshooting: Common Issues

**1. Extra blank lines in output**
- **Problem**: File opened without `newline=''`
- **Solution**: Always use `newline=''` when opening files for CSV

**2. Unicode/encoding errors**
- **Problem**: Special characters in data
- **Solution**: Specify encoding: `open('file.csv', 'r', encoding='utf-8')`

**3. Type errors when doing calculations**
- **Problem**: CSV values are always strings!
- **Solution**: Convert to int()/float() before calculations

## Data Processing Workflows

Now let's put it all together with real-world data processing tasks.

### Filtering Data

Filter students by grade:

````python
import csv

top_students = []
with open('data_files/students.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        if row['Grade'] in ['A', 'B']:
            top_students.append(row)

print('Found', len(top_students), 'top students')
for student in top_students:
    print(student['Name'] + ': Grade', student['Grade'])
````

### Transforming Data

Add a "Pass/Fail" column based on grade:

````python
import csv

students_with_status = []
with open('data_files/students.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        row['Status'] = 'Pass' if row['Grade'] in ['A', 'B', 'C'] else 'Fail'
        students_with_status.append(row)

# Write to new file
with open('data_files/students_with_status.csv', 'w', newline='') as f:
    fieldnames = ['Name', 'Age', 'Grade', 'Status']
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(students_with_status)

print("Transformed data written to data_files/students_with_status.csv")
````

### Joining Data from Multiple Files

Combine `data_files/students.csv` with `data_files/grades.csv` to get numeric scores:

````python
import csv

# Read grades into a lookup dictionary
grade_scores = {}
with open('data_files/grades.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        grade_scores[row['Grade']] = int(row['Score'])

# Add scores to student data
students_with_scores = []
with open('data_files/students.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        grade = row['Grade']
        # Check if grade exists in grade_scores, otherwise use 0
        if grade in grade_scores:
            row['Score'] = grade_scores[grade]
        else:
            row['Score'] = 0
        students_with_scores.append(row)

# Write combined data
with open('data_files/students_with_scores.csv', 'w', newline='') as f:
    fieldnames = ['Name', 'Age', 'Grade', 'Score']
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(students_with_scores)

print("Combined data written to data_files/students_with_scores.csv")
````

## Good Programming Practices

### 1. Always Use Context Managers (`with` statements)

Reference Lesson 9a: Always use `with open(...)` when working with files. This ensures files are properly closed, even if an error occurs.

### 2. Type Conversion is Your Responsibility

The CSV module doesn't auto-convert types. Always validate and convert data:
- `int(row['Age'])` for integers
- `float(row['Price'])` for decimals
- Handle `ValueError` if conversion fails

### 3. Handle Errors Gracefully

Reference Lesson 10: Use `try-except` for file operations:
- `FileNotFoundError` for missing files
- `ValueError` for type conversion errors
- Validate data before using it

### 4. Choose the Right Tool

- **Use `csv.reader/writer`** for files without headers or when performance is critical
- **Use `csv.DictReader/DictWriter`** for files with headers - more readable and less error-prone

Remember: Dictionaries are more readable but slightly slower than lists.
