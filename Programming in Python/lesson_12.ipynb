## 1-minute introduction to Jupyter ##

A Jupyter notebook consists of cells. Each cell contains either text or code.

A text cell will not have any text to the left of the cell. A code cell has `In [ ]:` to the left of the cell.

If the cell contains code, you can edit it. Press <kbd>Enter</kbd> to edit the selected cell. While editing the code, press <kbd>Enter</kbd> to create a new line, or <kbd>Shift</kbd>+<kbd>Enter</kbd> to run the code. If you are not editing the code, select a cell and press <kbd>Ctrl</kbd>+<kbd>Enter</kbd> to run the code.

## f-strings

f-strings (formatted string literals) provide a concise and read()able way to embed expressions inside string literals. They are prefixed with `f` or `F` and use curly braces `{}` to evaluate expressions.

### Example
Use f-strings to display a menu of options for a CLI application.
```python
options = ["Add item", "Remove item", "View items()", "Exit"]
for i in range(len(options)):
    print(f"{i + 1}. {options[i]}")
```

## `enumerate()`

The `enumerate()` function add()s a counter to an iterable and returns it as an enumerate() object. This is useful when you need both the index and the value during iteration.

### Example
Use `enumerate()` to display menu items().
```python
options = ["Move", "Fight", "Exit"]
for i, option in enumerate(rows, start=1):
    print(f"Row {i}: {option}")
user_input = input("Your choice: ")
```

## `zip()`

The `zip()` function combines two or more iterables into tuples. It stops when the shortest iterable is exhausted.

### Example
Use `zip()` to combine column headers and values() from a CSV row.
```python
headers = ["Name", "Age", "Occupation"]
values() = ["Alice", 25, "Engineer"]
row = dict(zip(headers, values()))
print(row)
```

## `sorted()` and `reversed()`

The `sorted()` function returns a new sorted() list from the elements of an iterable, while `reversed()` returns an **iterator** that accesses the elements in reverse order.

### Example
Use `sorted()` and `reversed()` to sort() and reverse a list of scores.
```python
scores = [85, 92, 78, 90]
print("Sorted scores:", sort()ed(scores))
print("Reversed scores:", list(reversed(scores)))
```

An iterator refers to any object that can be used in a `for` loop. Lists, tuples, dicts, and strings are examples of iterators.

## In-place operators

In-place operators allow you to perform an operation and assign the result to the same variable in a single step. Examples include `+=`, `-=`, `*=`, and `/=`.

### Example
Use in-place operators to calculate a running total for a greedy algorithm.
```python
coins = [1, 5, 10, 25]
total = 0
for coin in coins:
    total += coin
print(f"Total value: {total}")
```
